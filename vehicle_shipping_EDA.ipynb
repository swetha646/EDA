📌 1.Introduction
</h2>

The vehicle shipping industry plays an important role in global transportation and international trade by enabling vehicles to be moved efficiently between different countries, ports, and destinations. Every vehicle shipment generates a significant amount of data related to the vehicle, transportation route, shipping company, transportation mode, distance, weight, shipping cost, delivery duration, and shipment status.

This project focuses on performing Exploratory Data Analysis (EDA) on a large vehicle shipping dataset to understand the different patterns and trends associated with vehicle transportation. The dataset contains information about various vehicle makes, models, types, fuel types, manufacturing years, shipment years, origin and destination locations, shipping companies, shipment modes, shipping distances, vehicle weights, shipping costs, vehicle prices, delivery times, and delivery statuses.

The dataset contains both categorical and numerical variables, which allows different aspects of vehicle shipping operations to be examined. Categorical variables can be used to study shipment patterns across vehicle types, fuel types, shipping companies, countries, ports, transportation modes, and delivery statuses. Numerical variables can be analyzed to understand shipping costs, vehicle prices, distances, weights, and delivery durations.

The analysis begins with data understanding and data-quality analysis, where the structure of the dataset, column names, data types, unique values, missing values, duplicate records, and possible invalid or unusual values are examined. Since the dataset has already been cleaned, the cleaned data is then verified before proceeding with the main analysis.

The project then performs exploratory analysis to investigate vehicle characteristics, shipping costs, transportation routes, shipping companies, shipment modes, delivery performance, and time-based shipment patterns. Different aggregation and comparison techniques are used to identify important patterns within the data.

# 🔑 KEY FINDINGS / INSIGHTS

The Exploratory Data Analysis of **2.15 million vehicle shipment records** revealed the following key insights:

### 🚗 Vehicle Insights
- **Sedan** is the most frequently shipped vehicle type, with **530,701 shipments**.
- **Petrol** is the most common fuel type, accounting for **901,574 shipments**.
- **BMW** has the highest number of shipments among vehicle makes, with **162,735 shipments**.
- **C-Class** is the most frequently shipped vehicle model, with **67,881 shipments**.
- The average vehicle price is approximately **$20,310**.

### 🌍 Shipping & Route Insights
- The dataset covers shipments across **57 origin countries and 57 destination countries**.
- **UAE** has the highest number of shipments by origin country, with **163,554 shipments**.
- **USA** has the highest number of shipments by destination country, with **162,862 shipments**.
- **Jebel Ali** is the most frequently used origin port, while **Durban** is the most frequently used destination port.
- **Sea** is the most commonly used shipment mode, with **938,116 shipments**.

### 💰 Cost & Distance Insights
- The average shipping distance is approximately **7,499 km**.
- The average shipping cost is approximately **$4,676**.
- The average vehicle shipping cost represents approximately **29.22% of the vehicle price**.
- Shipping cost shows a **moderate positive correlation with distance (0.372)**.
- The relationship between **vehicle weight and shipping cost is very weak (correlation = 0.022)**.
- Shipping cost increases substantially across the distance ranges, indicating that **distance is an important factor associated with shipping cost**.

### ⏱️ Delivery Insights
- The average delivery time is approximately **19.72 days**, while the median is **15.5 days**.
- Distance and delivery time have a **positive correlation of 0.656**, indicating that longer shipping distances are generally associated with longer delivery times.
- The delivery status distribution shows that **delivered shipments form the largest group**, followed by delayed and in-transit shipments.
- Delivery time varies across shipping companies, shipment modes, vehicle types, and destination countries.

### ⭐ Customer, Insurance & Damage Insights
- The average customer rating is approximately **3.00 out of 5**.
- The correlation between delivery time and customer rating is almost zero (**-0.0003**), indicating no meaningful linear relationship in this dataset.
- The correlation between vehicle price and shipping cost is also almost zero (**0.0002**).
- Approximately **58.22%** of records indicate that insurance is included, while approximately **38.77%** indicate that it is not.
- Approximately **7.78%** of records report damage, while the majority report no damage.

### 📅 Time-Based Insights
- Shipment volumes remain relatively consistent across the years **2015–2024**, with no major variation in the number of shipments.
- **2018** recorded the highest shipment volume with **215,662 shipments**.
- **March** had the highest number of shipments, with **179,800 shipments**.
- Average shipping cost and average delivery time remain relatively stable across the shipment years.

### 📊 Overall Insight
Overall, the analysis shows that **shipping distance is one of the more important factors associated with both shipping cost and delivery time**. Vehicle characteristics such as price and weight show very weak linear relationships with shipping cost. The dataset also indicates stable shipment volumes over time, with sea transportation being the dominant shipping mode.
