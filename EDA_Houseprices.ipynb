# **Exploratory Data Analysis of House Prices**

This dataset contains information about houses, including their prices, number of bedrooms and bathrooms, living area, lot area, floors, waterfront view, condition, location, year built, and renovation details. The purpose of this analysis is to explore the dataset, identify patterns and relationships between house characteristics and house prices, detect data-quality issues and outliers, and derive meaningful insights.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Column Description

- **date** – The date on which the house was sold.
- **price** – The selling price of the house.
- **bedrooms** – The number of bedrooms.
- **bathrooms** – The number of bathrooms.
- **sqft_living** –The square feet of living area.
- **sqft_lot** – The total area of the property in square feet.
- **floors** – The number of floors.
- **waterfront** – If the house has a waterfront view.
- **view** – The quality of the view from the house.
- **condition** – The overall condition of the house.
- **sqft_above** – The area of the house above ground level in square feet.
- **sqft_basement** – The area of the basement in square feet.
- **yr_built** – The year in which the house was built.
- **yr_renovated** – The year in which the house was last renovated.
- **street** – The street address of the house.
- **city** – The city where the house is located.
- **statezip** – The state and ZIP code of the house.
- **country** – The country where the house is located.

```python
df['bathrooms'] =df['bathrooms'].astype(int)
```

```python
df['floors'] =df['floors'].astype(int)
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
```

# **Bivariate Analysis**

# Scatter plot

```python
sns.scatterplot(x ="sqft_living", y="price",data=df)
plt.title("Living Area VS Price")
plt.xlabel("living Area")
plt.ylabel("price")
plt.show()
```

```python
sns.scatterplot(x ="sqft_lot", y="price",data=df)
plt.title("Lot Area VS Price")
plt.xlabel("lot Area")
plt.ylabel("price")
plt.show()
```

# Bar plot

```python
bedroom_price = df[df['bedrooms'] >= 1].groupby('bedrooms')['price'].mean()
plt.figure(figsize=(8,5))
plt.bar(bedroom_price.index, bedroom_price.values)
plt.title('Average House Price by Number of Bedrooms')
plt.xlabel('Number of Bedrooms')
plt.ylabel('Average Price')
plt.show()
```

```python
city_price = df.groupby('city')['price'].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,5))
plt.bar(city_price.index, city_price.values)
plt.title('Average House Price by City')
plt.xlabel('City')
plt.ylabel('Average Price')
plt.xticks(rotation=45)

plt.show()
```
